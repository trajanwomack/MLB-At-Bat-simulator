## Specialist measurements
The original catcher, catch probability, jump, and running exports include blank rows. Missing values are not zeros.

```python
specialist_files = ['catcher_poptime', 'catch_probability', 'jump', 'running']
summary = []
for name in specialist_files:
    frame = pd.read_csv(ROOT / 'datasets/2026/hitters' / f'{name}.csv')
    metrics = frame.drop(columns=['last_name, first_name', 'player_id', 'year'])
    summary.append({'table': name, 'source_rows': len(frame), 'rows_with_measurements': int(metrics.notna().any(axis=1).sum())})
pd.DataFrame(summary)
```
